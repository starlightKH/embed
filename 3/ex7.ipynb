```python
import os

search_root = "/home/rbgus/test"    # 탐색을 시작할 상위 폴더
module_name = "dog.py"       # 찾고 싶은 모듈 파일명

found_paths = []

for root, dirs, files in os.walk(search_root):
    if module_name in files:
        full_path = os.path.join(root, module_name)
        found_paths.append(os.path.abspath(full_path))

if found_paths:
    print("찾은 모듈 위치들:")
    for path in found_paths:
        print(path)
else:
    print("모듈을 찾지 못했습니다.")
```
