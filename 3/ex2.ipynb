```python
import os

search_root = "/home/rbgus"    # 탐색을 시작할 상위 폴더
module_name = "game.py"       # 찾고 싶은 모듈 파일명

found_paths = []

for root, dirs, files in os.walk(search_root):
    if module_name in files:
        full_path = os.path.join(root)
        found_paths.append(os.path.abspath(full_path))
gamepath = found_paths[0]

print(gamepath)
```

```python
import sys
sys.path.append(gamepath)
print("\n추가 후 sys.path:")
for path in sys.path:
    print(path)
```
