```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])
b = np.array([1, 2, 3, 4, 5])

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print()

print("a + 100 =", a + 100)
print("a * 2 =", a * 2)
print("a / 10 =", a / 10)
print()

print("sum =", np.sum(a))
print("mean =", np.mean(a))
print("max =", np.max(a))
print("min =", np.min(a))
print()

print("a > 25 =", a > 25)
print("a[a > 25] =", a[a > 25])
print()

c = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("c =")
print(c)
print("(shape) =", c.shape)
print("(ndim) =", c.ndim)
print("(size) =", c.size)
print()
```
