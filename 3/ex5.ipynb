```python
import os

def main():
    current_dir = os.getcwd()
    items = os.listdir(current_dir)

    print("현재 작업 디렉토리")
    print(current_dir)
    print()

    print("[폴더 목록]")
    has_dir = False
    for item in items:
        full_path = os.path.join(current_dir, item)
        if os.path.isdir(full_path):
            print(item)
            has_dir = True

    if not has_dir:
        print("폴더가 없습니다.")

    print()
    print("[파일 목록]")
    has_file = False
    for item in items:
        full_path = os.path.join(current_dir, item)
        if os.path.isfile(full_path):
            print(item)
            has_file = True

    if not has_file:
        print("파일이 없습니다.")

if __name__ == "__main__":
    main()
```
